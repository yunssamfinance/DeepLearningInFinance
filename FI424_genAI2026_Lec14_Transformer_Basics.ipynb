## What is Generative AI

**Generative AI** refers to a subset of artificial intelligence technologies and algorithms that can **generate new content**, such as text, images, music, and even video, which resemble human-like creations. **Generative Adversarial Networks (GANs)** and **Variational Autoencoders (VAEs)** are two of the most significant breakthroughs in Generative AI, which are based on deep learning principles.
**Generative Adversarial Networks (GANs):** Introduced by Ian Goodfellow and his colleagues in 2014, GANs consist of **two neural networks, the generator and the discriminator**, trained simultaneously in a competitive setting. The generator learns to produce indistinguishable content from real data, while the discriminator learns to differentiate between real and generated content. This **adversarial process** enhances the quality of the generated content over time.

**Variational Autoencoders (VAEs):** VAEs are another pivotal architecture in the field of Generative AI. They are designed to **encode input data into a lower-dimensional space** and then **reconstruct it back** to the original form. Through this process, VAEs learn to capture the essential characteristics of the data, allowing them to generate new data points with similar properties.

This development was crucial for understanding the context and nuances of language, as it enables the model to make connections between distant words in a sentence or document.

The "Attention Is All You Need" paper not only introduced the Transformer architecture but also laid the foundation for subsequent developments in NLP. Thanks to their scalability and efficiency, transformer models have been the basis for many advanced text-to-text models, including **BERT, GPT series, and T5**, among others. These models are trained on vast datasets and can perform various language tasks, from translation and summarization to question-answering and text generation, **without requiring task-specific architectures**. The Transformer's ability to handle sequential data **in parallel**, scalability, and effective handling of long-range dependencies have made it a **cornerstone of modern NLP** research and applications. # Module 14: How a Transformer Works - The Big Picture

This module provides a **plain-English walkthrough** of how a Transformer processes text and generates predictions. We will walk through a concrete **5-stage pipeline** using the example sentence **"rates will rise"** to predict the next word. This is the conceptual foundation. In Lectures 14a and 14b, we will dive inside each stage and build the **attention mechanism** and **positional encodings** from scratch with detailed numerical examples.


## What Problem Is the Transformer Solving?

A language model has one job: given a sequence of words, **produce a probability distribution over what word comes next**.

The key challenge is **context**. "Bank" means something entirely different in "river bank" versus "central bank." The same word carries **different semantic weight** depending on what surrounds it.

The core idea behind the Transformer is deceptively simple but profoundly effective: Instead of reading left-to-right one word at a time (like a Recurrent Neural Network), the Transformer **looks at all words simultaneously** and learns which ones should influence each other.

If you are familiar with portfolio management in finance, think of it this way: each word is an asset. **Attention is the portfolio weight matrix**, deciding how much each word should invest in every other word. This lets the model learn that "rates" and "rise" are **strongly correlated**, while "will" (a mere auxiliary verb) should carry less weight in interpreting "rise."


## The Five-Stage Pipeline

Every Transformer follows the same **five-stage process**:

1. **Stage 1: Tokenize and Embed** - Convert words to integers, then look them up in a **learned embedding table**
2. **Stage 2: Add Positional Signal** - Inject information about **word order** into the embeddings
3. **Stage 3: Attention Layers** - Let each word **attend to all other words**, learning which connections matter
4. **Stage 4: Feed-Forward Layers** - Apply a small neural network to **refine each word representation independently**
5. **Stage 5: Output Head** - Project the final representation of the last token onto a **probability distribution over the vocabulary**

Critical note: Stages 3 and 4 are **stacked repeatedly**. GPT-2 stacks them **12 times**. GPT-4 is estimated to stack them roughly **96 times**. Each pass refines the representations, with **early layers catching syntax** and **later layers encoding semantics** and world knowledge.




*Figure 1: The five-stage Transformer pipeline. Teal = distinctive to Transformers. Stages ③ and ④ are stacked N times (GPT-2: 12×; GPT-4: ~96×).*

## Stage 1: Tokenize and Embed

Let us start with the example sentence: **"rates will rise"** and we want to predict the next word.

First, each word is mapped to an **integer ID** (tokenization), then looked up in an **embedding matrix**. The embedding matrix is a **learned parameter** that the model updates during training.

For readability, we use **d=4** (embedding dimension). In practice, GPT-3 uses **d=12,288**. This is one of the key scaling factors: **larger d means richer representations** but higher computational cost.




*Figure 2: Embedding lookup table. Each token becomes a 4-dimensional vector. These values are learned parameters — not hand-designed.*

```python
import numpy as np

# Define the embedding matrix (learned during training)
# Each row is a d-dimensional vector for a word
# In reality, d=12,288 for GPT-3. Here d=4 for readability.

embedding_matrix = {
    'rates': np.array([0.9, 0.1, 0.4, 0.7]),
    'will':  np.array([0.2, 0.8, 0.3, 0.5]),
    'rise':  np.array([0.6, 0.2, 0.9, 0.1])
}

print("Stage 1: Tokenize and Embed")
print("="*60)
print("\nEmbedding matrix (learned parameters):")
for word, vec in embedding_matrix.items():
    print(f"{word:10s}: {vec}")

print("\nThese embeddings capture the semantic meaning of each word.")
print("They are learned during training, not hand-designed.")
```

## Stage 2: Add Positional Signal

Embeddings alone **do not carry information about word order**. "rates will rise" means something very different from "rise will rates."

To fix this, the Transformer adds a **positional encoding** to each embedding. This vector captures information about **the position of the word** in the sequence.

We **add** the positional encoding to each embedding to create the **input representation** for the attention layers.




*Figure 3: Positional encoding addition. Each token embedding receives a small position-dependent offset. The = Input column is what enters the attention layers.*

```python
import pandas as pd

print("Stage 2: Add Positional Signal")
print("="*60)

# Positional encodings for each position
# These are learned during training (in some models) or computed via sinusoids (in the original Transformer)
# Details in Lecture 14b. For now, we just show the values.

pos_encodings = {
    'rates': np.array([0.0, 0.1, 0.0, 0.1]),  # position 1
    'will':  np.array([0.1, 0.0, 0.1, 0.0]),  # position 2
    'rise':  np.array([0.0, 0.1, 0.0, 0.1])   # position 3
}

# Add positional encoding to embedding
inputs = {}
for word in embedding_matrix.keys():
    inputs[word] = embedding_matrix[word] + pos_encodings[word]

# Display as a table
data = []
for word in ['rates', 'will', 'rise']:
    data.append({
        'token (position)': f"{word} ({list(embedding_matrix.keys()).index(word) + 1})",
        'embedding': str(embedding_matrix[word]),
        '+ pos.enc.': str(pos_encodings[word]),
        '= input': str(inputs[word])
    })

df = pd.DataFrame(data)
print("\nCombining embeddings with positional encodings:")
for word in ['rates', 'will', 'rise']:
    idx = list(embedding_matrix.keys()).index(word) + 1
    print(f"\n{word} (position {idx}):")
    print(f"  Embedding:       {embedding_matrix[word]}")
    print(f"  + Positional:    {pos_encodings[word]}")
    print(f"  = Input vector:  {inputs[word]}")

print("\n" + "="*60)
print("Note: We go deep into how positional encodings are constructed")
print("(sinusoidal vs. learned) in Lecture 14b.")
```

## Stage 3: Attention Layers - The Core of the Transformer

This is where the magic happens. Each attention layer produces a **new representation for every token** that blends information from all other tokens, **weighted by relevance**.

Intuition: The word "rise" should **pay close attention to "rates"** (they are economically linked), **less attention to "will"** (just a helping verb), and moderate attention to itself.

The attention mechanism computes a **weight matrix** showing how much each token should attend to every other token. In this example, we show a single **3x3 attention weight matrix**:




*Figure 4: Left — full attention weight matrix (darker = higher weight). Right — how "rise" distributes its attention. "rise" puts 55% weight on "rates" — economically sensible, since interest rates drive whether they rise.*

```python
import numpy as np

print("Stage 3: Attention Layers")
print("="*60)

# Attention weight matrix
# Rows: which token is "looking" (query)
# Columns: which tokens it attends to (keys)
# Values: normalized weights (sum to 1 across each row)

attention_weights = np.array([
    [0.60, 0.25, 0.15],  # rates attends: 60% to itself, 25% to will, 15% to rise
    [0.20, 0.50, 0.30],  # will attends:  20% to rates, 50% to itself, 30% to rise
    [0.55, 0.10, 0.35]   # rise attends:  55% to rates, 10% to will, 35% to itself
])

print("\nAttention Weight Matrix (3 tokens x 3 tokens):")
print("\n         rates  will  rise")
tokens = ['rates', 'will', 'rise']
for i, token in enumerate(tokens):
    print(f"{token:6s}  {attention_weights[i, 0]:.2f}  {attention_weights[i, 1]:.2f}  {attention_weights[i, 2]:.2f}")

print("\n" + "="*60)
print("Interpretation:")
print("- 'rates' puts 60% weight on itself, 25% on 'will', 15% on 'rise'")
print("- 'will'  puts 50% weight on itself, 30% on 'rise', 20% on 'rates'")
print("- 'rise'  puts 55% weight on 'rates', only 10% on 'will', 35% on itself")
print()
print("Key insight: 'rise' puts the highest weight (55%) on 'rates'.")
print("This makes intuitive sense: interest rates and rising yields are")
print("economically linked. The model learned this relationship.")
print()
print("After attention, each token's vector is no longer just its own")
print("embedding. It is a context-aware blend of all tokens, weighted")
print("by relevance.")
print()
print("Note: We cover the Q/K/V mechanics and how this attention matrix")
print("is computed in detail in Lecture 14a.")
```

```python
# For completeness, let us compute what the new representations look like
# after attention. This is a simplified version (in reality, there are more steps).

print("\nSimplified example: Blending via attention")
print("="*60)

# Original input vectors (from Stage 2)
V = np.array([
    inputs['rates'],
    inputs['will'],
    inputs['rise']
])

print("\nOriginal input vectors (V):")
for i, word in enumerate(['rates', 'will', 'rise']):
    print(f"{word:6s}: {V[i]}")

# Apply attention: new representation = attention_weights @ V
output_after_attention = attention_weights @ V

print("\nAfter attention (A @ V):")
for i, word in enumerate(['rates', 'will', 'rise']):
    print(f"{word:6s}: {output_after_attention[i]}")

print("\nNotice: Each token representation is now a weighted blend")
print("of the original tokens, based on the learned attention weights.")
```

## Stage 4: Feed-Forward Layer

After attention mixes information across tokens, a small **feed-forward network** processes each token **independently**. Think of this as a **per-token "interpretation" step**.

**No cross-token communication** happens here. It is purely pointwise. The **same feed-forward network** is applied to each token vector.

Together, Stages 3 and 4 form one **"Transformer layer."** A real Transformer **stacks this pair many times**. Each pass refines the representations. **Early layers catch syntax and grammar. Later layers encode semantics and world knowledge.**

For example, the first layer might learn that "will" is an auxiliary verb and adjust its weight accordingly. By the 10th layer, the model understands that **rising rates affect bond prices, equity valuations, and mortgage demand**.


```python
import numpy as np

print("Stage 4: Feed-Forward Layer")
print("="*60)

print("\nAfter attention, a feed-forward network (two linear layers")
print("with a nonlinearity like ReLU between them) processes each token.")

# Simple example: a feed-forward network that multiplies by a 4x4 matrix,
# applies ReLU, then multiplies by another 4x4 matrix to map back to d=4.

W1 = np.random.randn(4, 8) * 0.1  # First layer: d -> 2d
W2 = np.random.randn(8, 4) * 0.1  # Second layer: 2d -> d

def feedforward(x):
    # x is a 4-dimensional vector
    h = np.dot(x, W1)       # Project to hidden dimension
    h = np.maximum(h, 0)    # ReLU activation
    y = np.dot(h, W2)       # Project back to original dimension
    return y

print("\nApplying feed-forward to each token (after attention):")
output_after_ff = np.array([feedforward(output_after_attention[i]) for i in range(3)])

for i, word in enumerate(['rates', 'will', 'rise']):
    print(f"\n{word}:")
    print(f"  After attention:     {output_after_attention[i]}")
    print(f"  After feed-forward:  {output_after_ff[i]}")

print("\n" + "="*60)
print("Key point: Stages 3 (attention) and 4 (feed-forward) are stacked.")
print("- GPT-2: 12 layers")
print("- GPT-3: 96 layers")
print("- Each layer refines the representations further.")
```

## Stage 5: Output Head - Predicting the Next Word

We take the **final hidden state of the last token** ("rise") and project it onto the full vocabulary using a **linear layer**. Then we apply **softmax** to convert the raw scores into **probabilities**.

The model then returns a **probability distribution**: the probability that the next word is "sharply," "again," "in," etc.

To pick the actual next word, we can use **greedy decoding** (always pick highest probability), **temperature sampling** (higher temperature means more creative), or **beam search** (keep top-k candidates and continue from each).




*Figure 5: Softmax output over a 6-word vocabulary. "sharply" wins with 42% probability — the most likely next word given "rates will rise".*

```python
import numpy as np

print("Stage 5: Output Head - Predicting the Next Word")
print("="*60)

# Take the representation of the last token ("rise") after all layers
last_token_hidden = output_after_ff[2]  # 4-dimensional vector

print(f"\nLast token ('rise') hidden state: {last_token_hidden}")

# Project onto vocabulary (simplified: we show just 6 candidate words)
# In reality, vocabulary size is approximately 50,000 for GPT

vocab_scores = np.array([
    4.2,   # sharply
    3.1,   # again
    2.8,   # in
    2.7,   # further
    1.9,   # the
    0.4    # not
])

candidate_words = ['sharply', 'again', 'in', 'further', 'the', 'not']

# Apply softmax to convert scores to probabilities
def softmax(scores):
    exp_scores = np.exp(scores - np.max(scores))  # Subtract max for numerical stability
    return exp_scores / np.sum(exp_scores)

probs = softmax(vocab_scores)

print("\nLinear projection of last token onto vocabulary (raw logits):")
for word, score in zip(candidate_words, vocab_scores):
    print(f"  {word:10s}: {score:.1f}")

print("\nAfter softmax (probabilities):")
for word, prob in zip(candidate_words, probs):
    print(f"  {word:10s}: {prob:.4f} ({prob*100:.1f}%)")

print(f"\nSum of probabilities: {np.sum(probs):.4f} (should be 1.0)")

print("\n" + "="*60)
print("How to pick the next word:")
print("\n1. Greedy decoding: Always pick the highest probability.")
print(f"   Next word: '{candidate_words[np.argmax(probs)]}' (prob={probs[np.argmax(probs)]:.1%})")
print()
print("2. Temperature sampling: Scale logits by 1/T before softmax.")
print("   Higher T (e.g., 1.0) = more creative")
print("   Lower T (e.g., 0.1) = more deterministic")
print()
print("3. Beam search: Keep top-k candidates and continue from each.")
print("   k=3: continue from 'sharply', 'again', 'in'")
```

## Why the Transformer Replaced RNNs in Finance

The Transformer has three massive practical advantages for analyzing financial text and time-series:

**1. Long-Range Dependencies**
An RNN processing a 10-K filing must carry information about "revenue recognition policy" all the way to "net income adjustment" 500 words later, **compressing everything into a single hidden state**. This is a bottleneck. **Attention accesses any position directly in one step.** For financial document analysis, this is game-changing.

**2. Parallelism**
**All tokens are processed simultaneously.** This is why Transformers can be trained on billions of documents. GPUs can operate on the full attention matrix **in parallel** rather than sequentially, the way RNNs must. Training speed is **orders of magnitude faster**.

**3. Transfer Learning**
A model pre-trained on general text (FinBERT, BloombergGPT) already **encodes financial language patterns**. Fine-tuning on a small labeled dataset (e.g., earnings call sentiment, loan default risk) achieves **strong performance with far less labeled data** than training from scratch. This is especially valuable in finance where labeled datasets are expensive and hard to come by.


```python
print("Why Transformers Replaced RNNs in Finance")
print("="*80)

advantages = [
    {
        'Advantage': 'Long-Range Dependencies',
        'RNN Problem': '500-word gap compresses into one state',
        'Transformer Solution': 'Direct access to any position via attention',
        'Finance Impact': 'Analyze 10-K filings with full context'
    },
    {
        'Advantage': 'Parallelism',
        'RNN Problem': 'Sequential processing (one word at a time)',
        'Transformer Solution': 'All tokens processed in parallel',
        'Finance Impact': 'Train on billions of documents in weeks'
    },
    {
        'Advantage': 'Transfer Learning',
        'RNN Problem': 'Limited pre-training; language-specific',
        'Transformer Solution': 'Pre-trained on massive text corpora',
        'Finance Impact': 'Fine-tune FinBERT on small labeled dataset'
    }
]

import pandas as pd
df_advantages = pd.DataFrame(advantages)
print(df_advantages.to_string(index=False))

print("\n" + "="*80)
print("PRACTICAL EXAMPLE: Sentiment analysis of earnings calls")
print()
print("Task: Predict whether a stock will beat or miss EPS targets")
print("      based on the tone and content of the earnings call.")
print()
print("RNN approach:")
print("  - Collect labeled examples (expensive: analyst annotation)")
print("  - Train from scratch: weeks of GPU time")
print("  - Accuracy on small dataset: 72%")
print()
print("Transformer approach:")
print("  - Use FinBERT (pre-trained on financial text)")
print("  - Fine-tune on same small labeled dataset: hours of GPU time")
print("  - Accuracy: 84%")
print()
print("The Transformer leverages prior knowledge from large corpora.")
print("This is why Transformers dominate in finance today.")
```
